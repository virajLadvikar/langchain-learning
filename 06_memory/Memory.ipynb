## Memory In Langchain

```python
from langchain_groq import ChatGroq
from langgraph.checkpoint.memory import InMemorySaver
from dotenv import load_dotenv
load_dotenv()

llm = ChatGroq(
model="qwen/qwen3.8-27b"
)
```

### 3. Basic Example: Remember a Name

```python
from langchain.agents import create_agent
from langgraph.checkpoint.memory import InMemorySaver

# Create the agent
agents = create_agent(
    model=llm,
    checkpointer=InMemorySaver()
)

# Configure conversation thread
config = {
    "configurable": {
        "thread_id": "1"
    }
}

# First interaction
res = agents.invoke(
    {
        "messages": [
            {
                "role": "user",
                "content": "Hi, my name is Viraj"
            }
        ]
    },
    config=config
)

print(res["messages"][-1].content)
```

```python
res = agents.invoke({
    "messages":[
        {
            "role":"user",
            "content":"hi what is my name"
        
        }
    ]
},
config=config
)

print(res['messages'][-1].content)
```
